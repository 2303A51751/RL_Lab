```python

import gymnasium as gym
import numpy as np

# Create environment
env = gym.make("FrozenLake-v1")
print("Frozen Lake Environment is Created")

# Reset environment
state, info = env.reset()
print("state=", state)

# Display action space
print(env.action_space)

total_reward = 0
steps = 0

print("current_state:", state)

# Random agent
for i in range(100):

    action = env.action_space.sample()

    current_state = state

    next_state, reward, terminated, truncated, info = env.step(action)

    print(
        f"episode={i+1} | next_state={next_state} | "
        f"current_state={current_state} | reward={reward} | "
        f"terminated={terminated} | truncated={truncated} | "
        f"action={action}"
    )

    total_reward += reward
    steps += 1

    state = next_state

    if terminated or truncated:
        break

# Display results
print("total_reward=", total_reward)
print("average_reward=", total_reward / steps)

env.close()
```
